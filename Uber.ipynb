```python
### Simple categorizaion of the Resturant Types:

df_rest = df.copy()
df_rest['rest_type'] = df_rest['rest_type'].str.strip()
def categorize_rest_type(x):
    x_lower = x.lower()

    # Check for Pub/Bar/Lounge related terms
    is_bar_pub_lounge = any(word in x_lower for word in ['bar', 'pub', 'lounge', 'microbrewery', 'club', 'cocktail bar'])

    # Check for food-related terms
    is_food_related = any(word in x_lower for word in ['dining', 'cafe', 'quick bites', 'bakery', 'dessert', 'pizzeria', 'food court', 'bhojanalya', 'sweet shop', 'deli', 'fast food', 'meat shop', 'food truck', 'kiosk', 'mess', 'dhabha'])

    if is_bar_pub_lounge and is_food_related:
        return 'Pub/Bar/Lounge with Food'
    elif is_bar_pub_lounge:
        return 'Pub/Bar/Lounge (Drinks Only)'
    elif 'fine dining' in x_lower:
        return 'Fine Dining'
    elif 'casual dining' in x_lower:
        return 'Casual Dining'
    elif any(word in x_lower for word in ['quick bites', 'takeaway', 'delivery', 'food court', 'kiosk', 'mess', 'dhabha', 'bhojanalya', 'fast food']):
        return 'Quick Service'
    elif any(word in x_lower for word in ['cafe', 'bakery', 'dessert', 'sweet', 'beverage', 'pizzeria', 'irani cafe']):
        return 'Cafe & Dessert'
    else:
        return 'Other' # For any remaining categories not explicitly defined

df_rest['rest_category'] = df_rest['rest_type'].apply(categorize_rest_type)
print(df_rest['rest_category'].value_counts())
df_rest
```

Next, we'll create a Python file named `app.py` that will contain our Streamlit application. This app will connect to the `uber_eats.db` database and display the `restaurants` table.

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import sqlite3
import os
import matplotlib.pyplot as plt
import seaborn as sns

st.set_page_config(layout='wide')
st.title('Uber Eats Data Analysis')

# Function to load and preprocess restaurant data
@st.cache_data
def load_and_preprocess_restaurant_data(csv_path):
    df = pd.read_csv(csv_path)

    # Initial data cleaning and deduplication
    df = df.sort_values(by='name')
    df = df.drop_duplicates(subset='name', keep='first')

    # Column name standardization
    df.rename(columns={'approx_cost(for two people)': 'cost_for_two'}, inplace=True)
    df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
    df.rename(columns={'rate': 'customer_rating'}, inplace=True)

    # Clean 'name' column
    df['name'] = df['name'].astype(str).str.encode('latin1', errors='ignore').str.decode('utf-8', errors='ignore')
    df['name'] = df['name'].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True)
    df['name'] = df['name'].str.strip().str.title()

    # Clean 'customer_rating'
    df['customer_rating'] = df['customer_rating'].astype(str).str.replace('/5', '', regex=False)
    df['customer_rating'] = df['customer_rating'].replace(['NEW', '-', ''], '0')
    df['customer_rating'] = pd.to_numeric(df['customer_rating'], errors='coerce').fillna(0)

    # Clean 'cost_for_two'
    df['cost_for_two'] = df['cost_for_two'].astype(str).str.replace(',', '')
    df['cost_for_two'] = pd.to_numeric(df['cost_for_two'], errors='coerce').fillna(df['cost_for_two'].median()).astype(int)

    # Convert 'online_order' and 'book_table' to binary
    df['online_order'] = df['online_order'].astype(str).str.strip().str.title().map({'Yes': 1, 'No': 0})
    df['book_table'] = df['book_table'].astype(str).str.strip().str.title().map({'Yes': 1, 'No': 0})

    # Convert 'votes' to int
    df['votes'] = pd.to_numeric(df['votes'], errors='coerce').fillna(0).astype(int)

    # Clean 'location'
    df['location'] = df['location'].str.replace(r'\d+(st|nd|rd|th)\s*(Block|Cross)','',regex=True)
    df['location'] = df['location'].str.replace(r'\d+', '', regex=True)
    df['location'] = df['location'].str.strip().str.title()

    # Convert to category types
    df['location'] = df['location'].astype('category')
    df['cuisines'] = df['cuisines'].astype(str).str.strip().astype('category')

    return df

# Function to load and preprocess order data
@st.cache_data
def load_and_preprocess_order_data(json_path):
    df_orders = pd.read_json(json_path)

    # Convert 'discount_used' to binary (1/0)
    df_orders['discount_used'] = df_orders['discount_used'].map({'Yes': 1, 'No': 0}).astype(int)

    # Convert 'order_date' column to datetime objects
    df_orders['order_date'] = pd.to_datetime(df_orders['order_date'])

    # Extract month and year into a new column
    df_orders['order_month_year'] = df_orders['order_date'].dt.to_period('M')

    # Clean 'restaurant_name' column: strip whitespace and convert to title case
    df_orders['restaurant_name'] = df_orders['restaurant_name'].str.strip().str.title()

    return df_orders

# Database path
db_path = 'uber_eats.db'
restaurant_csv_path = '/content/Uber_Eats_data.csv'
orders_json_path = '/content/orders.json'

# Connect to SQLite database
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# Check if 'restaurants' table exists and populate if not
cursor.execute("SELECT name FROM sqlite_master WHERE type='table' AND name='restaurants';")
if cursor.fetchone() is None:
    st.info("Database table 'restaurants' not found. Loading and preprocessing restaurant data...")
    df_processed = load_and_preprocess_restaurant_data(restaurant_csv_path)
    df_processed.to_sql('restaurants', conn, if_exists='replace', index=False)

    # Create cuisines_table as well
    df_cuisine = df_processed.copy()
    df_cuisine['cuisines'] = df_cuisine['cuisines'].str.split(',')
    df_cuisine = df_cuisine.explode('cuisines')
    df_cuisine['cuisines'] = df_cuisine['cuisines'].str.strip().str.title()
    df_cuisine.to_sql('cuisines_table', conn, if_exists='replace', index=False)

    st.success("Database tables 'restaurants' and 'cuisines_table' created and populated.")
else:
    st.info("Database tables found. Loading data...")

# Load restaurant data from the 'restaurants' table (now guaranteed to exist)
df_restaurant_display = pd.read_sql_query('SELECT * FROM restaurants', conn)

# Load and preprocess order data
df_orders = load_and_preprocess_order_data(orders_json_path)

# --- Sidebar for Navigation ---
st.sidebar.title("Navigation")
page = st.sidebar.radio("Go to", ["Restaurant Data Insights", "Order Data Analysis"])

if page == "Restaurant Data Insights":
    st.header("Restaurant Data Insights")
    st.subheader("Raw Restaurant Data")
    st.dataframe(df_restaurant_display)

    # Query from cell 8cb65a44-3be7-466d-9fbc-2a118a2443b3: Which Bangalore locations have the highest average restaurant ratings?
    st.subheader("Top 10 Locations with Highest Average Restaurant Ratings")
    query_8cb65a44 = """
    SELECT location, AVG(customer_rating) AS avg_rating
    FROM restaurants
    GROUP BY location
    ORDER BY avg_rating DESC
    LIMIT 10;
    """
    df_query_8cb65a44 = pd.read_sql(query_8cb65a44, conn)
    st.dataframe(df_query_8cb65a44)

    # Query from cell 4292e757-daca-4940-8691-0a24c1a77427: Top 10 Restaurants which have the highest Votes ?
    st.subheader("Top 10 Restaurants with Highest Votes")
    query_4292e757 = """
    SELECT name, location, votes
    FROM restaurants
    ORDER BY votes DESC
    LIMIT 10;
    """
    df_query_4292e757 = pd.read_sql(query_4292e757, conn)
    st.dataframe(df_query_4292e757)

    # Query from cell 474be0f5-f757-40f9-a7cf-d41a22326303: List Top Locations which have the highly voted restaurants ?
    st.subheader("Top 10 Locations with Highly Voted Restaurants (Votes > 1000)")
    query_474be0f5 = """
    SELECT location, COUNT(*) AS high_vote_restaurants
    FROM restaurants
    WHERE votes > 1000
    GROUP BY location
    ORDER BY high_vote_restaurants DESC
    LIMIT 10;
    """
    df_query_474be0f5 = pd.read_sql(query_474be0f5, conn)
    st.dataframe(df_query_474be0f5)

    # Query from cell 7cfe8483-8209-4dca-a00a-4f4b810f9465: Which locations are over-saturated with restaurants?
    st.subheader("Top 10 Over-Saturated Locations with Restaurants")
    query_7cfe8483 = """
    SELECT location, COUNT(*) AS total_restaurants
    FROM restaurants
    GROUP BY location
    ORDER BY total_restaurants DESC
    LIMIT 10;
    """
    df_query_7cfe8483 = pd.read_sql(query_7cfe8483, conn)
    st.dataframe(df_query_7cfe8483)

    # Query from cell 0534c1d1-bbde-40ef-a6ab-6ba07a8a7b01: Which cuisines are most common in Bangalore? && Which cuisines receive the highest average ratings?
    st.subheader("Top 10 Most Common Cuisines and Their Average Ratings (>= 20 Restaurants)")
    query_0534c1d1 = """
    SELECT
        cuisines,
        COUNT(*) AS total_restaurants,
        ROUND(AVG(customer_rating), 2) AS avg_rating
    FROM cuisines_table
    GROUP BY cuisines
    HAVING total_restaurants >= 20
    ORDER BY total_restaurants DESC, avg_rating DESC
    LIMIT 10;
    """
    df_query_0534c1d1 = pd.read_sql(query_0534c1d1, conn)
    st.dataframe(df_query_0534c1d1)

    # Query from cell c42f9ec3-a071-4f19-94f8-d1e29a7700d1: Finding high performing Cuisines, Keeping a minimum limit of 50 restaurants of same cuisine and is highly rated (4 and above)
    st.subheader("Top High Performing Cuisines (>= 50 Restaurants, Rating >= 4)")
    query_c42f9ec3 = """
    SELECT
        cuisines,
        COUNT(*) AS total_restaurants,
        ROUND(AVG(customer_rating), 2) AS avg_rating
    FROM cuisines_table
    GROUP BY cuisines
    HAVING total_restaurants >= 50
    AND avg_rating >= 4
    ORDER BY avg_rating DESC, total_restaurants DESC
    LIMIT 10;
    """
    df_query_c42f9ec3 = pd.read_sql(query_c42f9ec3, conn)
    st.dataframe(df_query_c42f9ec3)

    # Display the results of the query from cell 5591c286-8d5c-460f-9a6d-dc696869b2f5 as a table
    st.subheader("Relationship Between Restaurant Cost and Rating (Top 15)")
    query_cell_55 = """
    SELECT
        location,
        cuisines,

        CASE
            WHEN cost_for_two < 300 THEN 'Low'
            WHEN cost_for_two BETWEEN 300 AND 700 THEN 'Mid'
            ELSE 'Premium'
        END AS price_category,

        COUNT(*) AS total_restaurants,
        ROUND(AVG(customer_rating), 2) AS avg_rating,
        ROUND(AVG(votes), 0) AS avg_votes

    FROM cuisines_table

    GROUP BY location, cuisines, price_category

    HAVING total_restaurants >= 10

    ORDER BY avg_rating DESC, avg_votes DESC

    LIMIT 15;
    """
    df_query_55 = pd.read_sql(query_cell_55, conn)
    st.dataframe(df_query_55)

    # Query for 'Which restaurant type has highest votes and does that affect online orders ??'
    st.subheader("Restaurant Type and Online Order Impact on Votes (Top 15)")
    query_cell_56 = """
    SELECT
        rest_type,
        online_order,
        COUNT(*) AS total_restaurants,
        SUM(votes) AS total_votes,
        ROUND(AVG(votes), 0) AS avg_votes

    FROM restaurants

    GROUP BY rest_type, online_order

    HAVING total_restaurants >= 10

    ORDER BY avg_votes DESC

    LIMIT 15;
    """
    df_query_56 = pd.read_sql(query_cell_56, conn)
    st.dataframe(df_query_56)

    # Query for 'Which locations show high demand but lower average ratings?'
    st.subheader("Locations with High Demand but Lower Average Ratings (Top 5)")
    query_cell_57 = """
    SELECT
        location,
        COUNT(*) AS total_restaurants,
        ROUND(AVG(votes), 0) AS avg_votes,
        ROUND(AVG(customer_rating), 2) AS avg_rating

    FROM restaurants

    GROUP BY location

    HAVING total_restaurants >= 30
    AND avg_rating < 3.75

    ORDER BY avg_votes DESC, avg_rating ASC

    LIMIT 5;
    """
    df_query_57 = pd.read_sql(query_cell_57, conn)
    st.dataframe(df_query_57)

    # Query for 'What combination of factors maximizes restaurant success on Uber Eats?'
    st.subheader("Factors Maximizing Restaurant Success on Uber Eats (Top 15)")
    query_cell_58 = """
    SELECT
        location,
        cuisines,

        CASE
            WHEN cost_for_two < 300 THEN 'Low'
            WHEN cost_for_two BETWEEN 300 AND 700 THEN 'Mid'
            ELSE 'Premium'
        END AS price_category,

        online_order,

        COUNT(*) AS total_restaurants,
        ROUND(AVG(customer_rating), 2) AS avg_rating,
        ROUND(AVG(votes), 0) AS avg_votes

    FROM cuisines_table

    GROUP BY location, cuisines, price_category, online_order

    HAVING total_restaurants >= 10

    ORDER BY avg_rating DESC, avg_votes DESC

    LIMIT 15;
    """
    df_query_58 = pd.read_sql(query_cell_58, conn)
    st.dataframe(df_query_58)

elif page == "Order Data Analysis":
    st.header("Order Data Analysis")
    st.subheader("Raw Order Data")
    st.dataframe(df_orders)

    # Top 5 Restaurants with the maximum Order Value
    st.subheader("Top 5 Restaurants by Total Order Value")
    top5_order_value = df_orders.groupby('restaurant_name')['order_value'].sum().nlargest(5)
    st.dataframe(top5_order_value)

    # Monthly sales and discount usage
    st.subheader("Monthly Sales and Discount Usage")
    monthly_sales_and_discounts = df_orders.groupby('order_month_year').agg(
        total_sales=('order_value', 'sum'),
        total_discounts_used=('discount_used', 'sum')
    ).sort_values(by='total_sales', ascending=False)
    st.dataframe(monthly_sales_and_discounts)

    # Total orders over the last 6 months
    st.subheader("Total Orders Over the Last 6 Months")
    latest_month = df_orders['order_month_year'].max()
    start_month = latest_month - 5 # Corrected fix: Direct subtraction for Period objects
    df_last_6_months = df_orders[df_orders['order_month_year'] >= start_month]
    orders_last_6_months = df_last_6_months.groupby('order_month_year').size().reset_index(name='total_orders')
    st.dataframe(orders_last_6_months)

    # Plotting total orders over the last 6 months
    fig_6_months_orders = plt.figure(figsize=(10, 6))
    sns.barplot(x='order_month_year', y='total_orders', data=orders_last_6_months, palette='viridis', hue='order_month_year', legend=False)
    plt.title('Total Orders Over the Last 6 Months')
    plt.xlabel('Month and Year')
    plt.ylabel('Total Orders')
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    st.pyplot(fig_6_months_orders)
    plt.close(fig_6_months_orders)

    # Average order value over the last 3 months
    st.subheader("Average Order Value Over the Last 3 Months")
    latest_month = df_orders['order_month_year'].max()
    start_month_3_months = latest_month - 2
    df_last_3_months = df_orders[df_orders['order_month_year'] >= start_month_3_months]
    average_order_value_last_3_months = df_last_3_months.groupby('order_month_year')['order_value'].mean().round(2).reset_index(name='average_order_value')
    st.dataframe(average_order_value_last_3_months)

    # Plotting average order value over the last 3 months
    fig_3_months_avg_value = plt.figure(figsize=(8, 5))
    sns.barplot(x='order_month_year', y='average_order_value', data=average_order_value_last_3_months, palette='magma', hue='order_month_year', legend=False)
    plt.title('Average Order Value Over the Last 3 Months')
    plt.xlabel('Month and Year')
    plt.ylabel('Average Order Value')
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    st.pyplot(fig_3_months_avg_value)
    plt.close(fig_3_months_avg_value)

# Close the database connection
conn.close()
```

### Adjusting Column Widths in Streamlit

To adjust column widths, you use `st.columns()` and pass a list of integers or fractions representing the relative widths. For example, `st.columns([1, 2, 1])` would create three columns where the second column is twice as wide as the first and third.

Here's an example of how to modify your `app.py` to use `st.columns` for better layout:

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import sqlite3
import os
import matplotlib.pyplot as plt
import seaborn as sns

st.set_page_config(layout='wide')
st.title('Uber Eats Data Analysis')

# Function to load and preprocess restaurant data
@st.cache_data
def load_and_preprocess_restaurant_data(csv_path):
    df = pd.read_csv(csv_path)

    # Initial data cleaning and deduplication
    df = df.sort_values(by='name')
    df = df.drop_duplicates(subset='name', keep='first')

    # Column name standardization
    df.rename(columns={'approx_cost(for two people)': 'cost_for_two'}, inplace=True)
    df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
    df.rename(columns={'rate': 'customer_rating'}, inplace=True)

    # Clean 'name' column
    df['name'] = df['name'].astype(str).str.encode('latin1', errors='ignore').str.decode('utf-8', errors='ignore')
    df['name'] = df['name'].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True)
    df['name'] = df['name'].str.strip().str.title()

    # Clean 'customer_rating'
    df['customer_rating'] = df['customer_rating'].astype(str).str.replace('/5', '', regex=False)
    df['customer_rating'] = df['customer_rating'].replace(['NEW', '-', ''], '0')
    df['customer_rating'] = pd.to_numeric(df['customer_rating'], errors='coerce').fillna(0).round(2)

    # Clean 'cost_for_two'
    df['cost_for_two'] = df['cost_for_two'].astype(str).str.replace(',', '')
    df['cost_for_two'] = pd.to_numeric(df['cost_for_two'], errors='coerce').fillna(df['cost_for_two'].median()).astype(int)

    # Convert 'online_order' and 'book_table' to binary
    df['online_order'] = df['online_order'].astype(str).str.strip().str.title().map({'Yes': 1, 'No': 0})
    df['book_table'] = df['book_table'].astype(str).str.strip().str.title().map({'Yes': 1, 'No': 0})

    # Convert 'votes' to int
    df['votes'] = pd.to_numeric(df['votes'], errors='coerce').fillna(0).astype(int)

    # Clean 'location'
    df['location'] = df['location'].str.replace(r'\d+(st|nd|rd|th)\s*(Block|Cross)','',regex=True)
    df['location'] = df['location'].str.replace(r'\d+', '', regex=True)
    df['location'] = df['location'].str.strip().str.title()

    # Convert to category types
    df['location'] = df['location'].astype('category')
    df['cuisines'] = df['cuisines'].astype(str).str.strip().astype('category')

    # Categorize restaurant type
    df['rest_type'] = df['rest_type'].str.strip()
    def categorize_rest_type(x):
        x_lower = x.lower()
        is_bar_pub_lounge = any(word in x_lower for word in ['bar', 'pub', 'lounge', 'microbrewery', 'club', 'cocktail bar'])
        is_food_related = any(word in x_lower for word in ['dining', 'cafe', 'quick bites', 'bakery', 'dessert', 'pizzeria', 'food court', 'bhojanalya', 'sweet shop', 'deli', 'fast food', 'meat shop', 'food truck', 'kiosk', 'mess', 'dhabha'])

        if is_bar_pub_lounge and is_food_related:
            return 'Pub/Bar/Lounge with Food'
        elif is_bar_pub_lounge:
            return 'Pub/Bar/Lounge (Drinks Only)'
        elif 'fine dining' in x_lower:
            return 'Fine Dining'
        elif 'casual dining' in x_lower:
            return 'Casual Dining'
        elif any(word in x_lower for word in ['quick bites', 'takeaway', 'delivery', 'food court', 'kiosk', 'mess', 'dhabha', 'bhojanalya', 'fast food']):
            return 'Quick Service'
        elif any(word in x_lower for word in ['cafe', 'bakery', 'dessert', 'sweet', 'beverage', 'pizzeria', 'irani cafe']):
            return 'Cafe & Dessert'
        else:
            return 'Other'
    df['rest_category'] = df['rest_type'].apply(categorize_rest_type)

    return df

# Function to load and preprocess order data
@st.cache_data
def load_and_preprocess_order_data(json_path):
    df_orders = pd.read_json(json_path)

    # Convert 'discount_used' to binary (1/0)
    df_orders['discount_used'] = df_orders['discount_used'].map({'Yes': 1, 'No': 0}).astype(int)

    # Convert 'order_date' column to datetime objects
    df_orders['order_date'] = pd.to_datetime(df_orders['order_date'])

    # Extract month and year into a new column
    df_orders['order_month_year'] = df_orders['order_date'].dt.to_period('M').dt.start_time

    # Clean 'restaurant_name' column: strip whitespace and convert to title case
    df_orders['restaurant_name'] = df_orders['restaurant_name'].str.strip().str.title()

    return df_orders

# Database path
db_path = 'uber_eats.db'
restaurant_csv_path = '/content/Uber_Eats_data.csv'
orders_json_path = '/content/orders.json'

# Connect to SQLite database
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# Check if 'restaurants' table exists and populate if not
cursor.execute("SELECT name FROM sqlite_master WHERE type='table' AND name='restaurants';")
if cursor.fetchone() is None:
    st.info("Database table 'restaurants' not found. Loading and preprocessing restaurant data...")
    df_processed = load_and_preprocess_restaurant_data(restaurant_csv_path)
    df_processed.to_sql('restaurants', conn, if_exists='replace', index=False)

    # Create cuisines_table as well
    df_cuisine = df_processed.copy()
    df_cuisine['cuisines'] = df_cuisine['cuisines'].str.split(',')
    df_cuisine = df_cuisine.explode('cuisines')
    df_cuisine['cuisines'] = df_cuisine['cuisines'].str.strip().str.title()
    df_cuisine.to_sql('cuisines_table', conn, if_exists='replace', index=False)

    st.success("Database tables 'restaurants' and 'cuisines_table' created and populated.")
else:
    st.info("Database tables found. Loading data...")

# Load restaurant data from the 'restaurants' table (now guaranteed to exist)
df_restaurant_display = pd.read_sql_query('SELECT * FROM restaurants', conn)

# Load and preprocess order data
df_orders = load_and_preprocess_order_data(orders_json_path)

# --- Sidebar for Navigation ---
st.sidebar.title("Navigation")
page = st.sidebar.radio("Go to", ["Restaurant Data Insights", "Order Data Analysis"])

if page == "Restaurant Data Insights":
    st.header("Restaurant Data Insights")
    st.subheader("Raw Restaurant Data")
    st.markdown("**Key Insights:**\n*   This table provides a raw view of the restaurant data after initial cleaning and preprocessing.\n*   It contains comprehensive information about each restaurant, including name, online order availability, booking options, ratings, votes, location, cuisine, and cost for two.")
    st.dataframe(
        df_restaurant_display,
        column_config={
            "name": st.column_config.TextColumn("Restaurant Name", width="large"),
            "phone": st.column_config.TextColumn("Phone", width="medium"),
            "location": st.column_config.TextColumn("Location", width="medium"),
            "rest_type": st.column_config.TextColumn("Restaurant Type", width="medium"),
            "dish_liked": st.column_config.TextColumn("Dish Liked", width="large"),
            "cuisines": st.column_config.TextColumn("Cuisines", width="large"),
            "listed_in(type)": st.column_config.TextColumn("Listed Type", width="medium"),
            "listed_in(city)": st.column_config.TextColumn("Listed City", width="medium")
        }
    )

    col1, col2 = st.columns([2, 1]) # Example of 2 columns with ratio 2:1

    with col1:
        # Query from cell 8cb65a44-3be7-446d-9fbc-2a118a2443b3: Which Bangalore locations have the highest average restaurant ratings?
        st.subheader("Top 10 Locations with Highest Average Restaurant Ratings")
        st.markdown("**Key Insights:**\n*   Identifies locations where restaurants generally receive higher customer ratings.\n*   Lavelle Road consistently has highly-rated restaurants, suggesting a strong culinary scene or discerning customer base.")
        query_8cb65a44 = """
        SELECT location, AVG(customer_rating) AS avg_rating
        FROM restaurants
        GROUP BY location
        ORDER BY avg_rating DESC
        LIMIT 10;
        """
        df_query_8cb65a44 = pd.read_sql(query_8cb65a44, conn)
        st.dataframe(
            df_query_8cb65a44,
            column_config={
                "location": st.column_config.TextColumn("Location", width="medium"),
                "avg_rating": st.column_config.NumberColumn(
                    "Average Rating",
                    format="%.3f"
                )
            }
        )

        # Query from cell 4292e757-daca-4940-8691-0a24c1a77427: Top 10 Restaurants which have the highest Votes ?
        st.subheader("Top 10 Restaurants with Highest Votes")
        st.markdown("**Key Insights:**\n*   Highlights individual restaurants that are highly popular and have received a large number of customer votes.\n*   These restaurants often indicate strong brand recognition, consistent quality, or effective marketing strategies.")
        query_4292e757 = """
        SELECT name, location, votes
        FROM restaurants
        ORDER BY votes DESC
        LIMIT 10;
        """
        df_query_4292e757 = pd.read_sql(query_4292e757, conn)
        st.dataframe(
            df_query_4292e757,
            column_config={
                "name": st.column_config.TextColumn("Restaurant Name", width="large"),
                "location": st.column_config.TextColumn("Location", width="medium")
            }
        )

        # Query from cell 474be0f5-f757-40f9-a7cf-d41a22326303: List Top Locations which have the highly voted restaurants ?
        st.subheader("Top 10 Locations with Highly Voted Restaurants (Votes > 1000)")
        st.markdown("**Key Insights:**\n*   Pinpoints areas with a concentration of popular restaurants, indicating high customer engagement and potential for business growth.\n*   Locations like Koramangala and Indiranagar are hubs for highly-voted establishments, suggesting high demand in these areas.")
        query_474be0f5 = """
        SELECT location, COUNT(*) AS high_vote_restaurants
        FROM restaurants
        WHERE votes > 1000
        GROUP BY location
        ORDER BY high_vote_restaurants DESC
        LIMIT 10;
        """
        df_query_474be0f5 = pd.read_sql(query_474be0f5, conn)
        st.dataframe(
            df_query_474be0f5,
            column_config={
                "location": st.column_config.TextColumn("Location", width="medium")
            }
        )

    with col2:
        # Query from cell 7cfe8483-8209-dca-a00a-4f4b810f9465: Which locations are over-saturated with restaurants?
        st.subheader("Top 10 Over-Saturated Locations with Restaurants")
        st.markdown("**Key Insights:**\n*   Identifies locations with the highest number of restaurants, which could indicate high competition but also high demand.\n*   Koramangala, Whitefield, and Indiranagar have a very large number of restaurants, suggesting these are key commercial and residential areas.")
        query_7cfe8483 = """
        SELECT location, COUNT(*) AS total_restaurants
        FROM restaurants
        GROUP BY location
        ORDER BY total_restaurants DESC
        LIMIT 10;
        """
        df_query_7cfe8483 = pd.read_sql(query_7cfe8483, conn)
        st.dataframe(
            df_query_7cfe8483,
            column_config={
                "location": st.column_config.TextColumn("Location", width="medium")
            }
        )

        # Query from cell 0534c1d1-bbde-40ef-a6ab-6ba07a8a7b01: Which cuisines are most common in Bangalore? && Which cuisines receive the highest average ratings?
        st.subheader("Top 10 Most Common Cuisines and Their Average Ratings (>= 20 Restaurants)")
        st.markdown("**Key Insights:**\n*   Reveals the most prevalent cuisines across Bangalore and their average customer ratings.\n*   North Indian and Chinese are the most common cuisines, indicating widespread popularity.")
        query_0534c1d1 = """
        SELECT
            cuisines,
            COUNT(*) AS total_restaurants,
            ROUND(AVG(customer_rating), 2) AS avg_rating
        FROM cuisines_table
        GROUP BY cuisines
        HAVING total_restaurants >= 20
        ORDER BY total_restaurants DESC, avg_rating DESC
        LIMIT 10;
        """
        df_query_0534c1d1 = pd.read_sql(query_0534c1d1, conn)
        st.dataframe(
            df_query_0534c1d1,
            column_config={
                "cuisines": st.column_config.TextColumn("Cuisines", width="large")
            }
        )

        # Query from cell c42f9ec3-a071-4f19-94f8-d1e29a7700d1: Finding high performing Cuisines, Keeping a minimum limit of 50 restaurants of same cuisine and is highly rated (4 and above)
        st.subheader("Top High Performing Cuisines (>= 50 Restaurants, Rating >= 4)")
        st.markdown("**Key Insights:**\n*   Highlights cuisines that are both widely available and highly rated, indicating strong customer satisfaction.\n*   Mediterranean, European, and Asian cuisines stand out as high-performing, suggesting demand for diverse and quality options.")
        query_c42f9ec3 = """
        SELECT
            cuisines,
            COUNT(*) AS total_restaurants,
            ROUND(AVG(customer_rating), 2) AS avg_rating
        FROM cuisines_table
        GROUP BY cuisines
        HAVING total_restaurants >= 50
        AND avg_rating >= 4
        ORDER BY avg_rating DESC, total_restaurants DESC
        LIMIT 10;
        """
        df_query_c42f9ec3 = pd.read_sql(query_c42f9ec3, conn)
        st.dataframe(
            df_query_c42f9ec3,
            column_config={
                "cuisines": st.column_config.TextColumn("Cuisines", width="large")
            }
        )

    # Display the results of the query from cell 5591c286-8d5c-460f-9a6d-dc696869b2f5 as a table
    st.subheader("Relationship Between Restaurant Cost and Rating (Top 15)")
    st.markdown("**Key Insights:**\n*   Analyzes how pricing (Low, Mid, Premium) impacts customer ratings and votes across different locations and cuisines.\n*   Premium restaurants in high-demand areas with specific cuisines (like Asian and Italian in Indiranagar and Koramangala) tend to achieve higher ratings and votes.")
    query_cell_55 = """
    SELECT
        location,
        cuisines,

        CASE
            WHEN cost_for_two < 300 THEN 'Low'
            WHEN cost_for_two BETWEEN 300 AND 700 THEN 'Mid'
            ELSE 'Premium'
        END AS price_category,

        COUNT(*) AS total_restaurants,
        ROUND(AVG(customer_rating), 2) AS avg_rating,
        ROUND(AVG(votes), 0) AS avg_votes

    FROM cuisines_table

    GROUP BY location, cuisines, price_category

    HAVING total_restaurants >= 10

    ORDER BY avg_rating DESC, avg_votes DESC

    LIMIT 15;
    """
    df_query_55 = pd.read_sql(query_cell_55, conn)
    st.dataframe(
        df_query_55,
        column_config={
            "location": st.column_config.TextColumn("Location", width="medium"),
            "cuisines": st.column_config.TextColumn("Cuisines", width="medium"),
            "price_category": st.column_config.TextColumn("Price Category", width="small"),
            "total_restaurants": st.column_config.NumberColumn("Total Restaurants", width="small"),
            "avg_rating": st.column_config.NumberColumn("Avg Rating", format="%.2f", width="small"),
            "avg_votes": st.column_config.NumberColumn("Avg Votes", format="%.0f", width="small")
        }
    )

    # Query for 'Which restaurant type has highest votes and does that affect online orders ??'
    st.subheader("Restaurant Category and Online Order Impact on Votes (Top 15)")
    st.markdown("**Key Insights:**\n*   Examines the relationship between restaurant category, online ordering availability, and customer engagement (votes).\n*   'Pub/Bar/Lounge (Drinks Only)' and 'Pub/Bar/Lounge with Food' categories tend to have the highest average votes, especially when online ordering is available, indicating strong popularity and customer interaction.")
    query_cell_56 = """
    SELECT
        rest_category,
        online_order,
        COUNT(*) AS total_restaurants,
        SUM(votes) AS total_votes,
        ROUND(AVG(votes), 0) AS avg_votes

    FROM restaurants

    GROUP BY rest_category, online_order

HAVING total_restaurants >= 10

    ORDER BY avg_votes DESC

    LIMIT 15;
    """
    df_query_56 = pd.read_sql(query_cell_56, conn)
    st.dataframe(
        df_query_56,
        column_config={
            "rest_category": st.column_config.TextColumn("Restaurant Category", width="large"),
            "online_order": st.column_config.NumberColumn("Online Order (1/0)", width="small"),
            "total_restaurants": st.column_config.NumberColumn("Total Restaurants", width="small"),
            "total_votes": st.column_config.NumberColumn("Total Votes", width="small"),
            "avg_votes": st.column_config.NumberColumn("Avg Votes", format="%.0f", width="small")
        }
    )

    # Query for 'Which locations show high demand but lower average ratings?'
    st.subheader("Locations with High Demand but Lower Average Ratings (Top 5)")
    st.markdown("**Key Insights:**\n*   Identifies areas where restaurants receive a high number of votes (high demand) but relatively lower average ratings.\n*   Locations like JP Nagar, Bannerghatta Road, and BTM show a gap where high customer interest isn't fully matched by high satisfaction, potentially indicating areas for improvement or new market entry.")
    query_cell_57 = """
    SELECT
        location,
        COUNT(*) AS total_restaurants,
        ROUND(AVG(votes), 0) AS avg_votes,
        ROUND(AVG(customer_rating), 2) AS avg_rating

    FROM restaurants

    GROUP BY location

    HAVING total_restaurants >= 30
    AND avg_rating < 3.75

    ORDER BY avg_votes DESC, avg_rating ASC

    LIMIT 5;
    """
    df_query_57 = pd.read_sql(query_cell_57, conn)
    st.dataframe(
        df_query_57,
        column_config={
            "location": st.column_config.TextColumn("Location", width="medium"),
            "total_restaurants": st.column_config.NumberColumn("Total Restaurants", width="small"),
            "avg_votes": st.column_config.NumberColumn("Avg Votes", format="%.0f", width="small"),
            "avg_rating": st.column_config.NumberColumn("Avg Rating", format="%.2f", width="small")
        }
    )

    # Query for 'What combination of factors maximizes restaurant success on Uber Eats?'
    st.subheader("Factors Maximizing Restaurant Success on Uber Eats (Top 15)")
    st.markdown("**Key Insights:**\n*   Examines the interplay of location, cuisine, price category, and online ordering to determine what drives restaurant success.\n*   Premium-priced restaurants with online ordering in prime locations like Lavelle Road and St. Marks Road, offering diverse cuisines, demonstrate high average ratings and votes.")
    query_cell_58 = """
    SELECT
        location,
        CASE
            WHEN cost_for_two < 300 THEN 'Low'
            WHEN cost_for_two BETWEEN 300 AND 700 THEN 'Mid'
            ELSE 'Premium'
        END AS price_category,

        online_order,

        COUNT(*) AS total_restaurants,
        ROUND(AVG(customer_rating), 2) AS avg_rating,
        ROUND(AVG(votes), 0) AS avg_votes

    FROM cuisines_table

    GROUP BY location, price_category, online_order

    HAVING total_restaurants >= 10

    ORDER BY avg_rating DESC, avg_votes DESC

    LIMIT 15;
    """
    df_query_58 = pd.read_sql(query_cell_58, conn)
    st.dataframe(
        df_query_58,
        column_config={
            "location": st.column_config.TextColumn("Location", width="medium"),
            "price_category": st.column_config.TextColumn("Price Category", width="small"),
            "online_order": st.column_config.NumberColumn("Online Order (1/0)", width="small"),
            "total_restaurants": st.column_config.NumberColumn("Total Restaurants", width="small"),
            "avg_rating": st.column_config.NumberColumn("Avg Rating", format="%.2f", width="small"),
            "avg_votes": st.column_config.NumberColumn("Avg Votes", format="%.0f", width="small")
        }
    )

elif page == "Order Data Analysis":
    st.header("Order Data Analysis")
    st.subheader("Raw Order Data")
    st.markdown("**Key Insights:**\n*   This table displays the raw order data after initial cleaning and preprocessing.\n*   It includes order ID, restaurant name, order date, order value, discount usage, and payment method.")
    st.dataframe(
        df_orders,
        column_config={
            "order_id": st.column_config.TextColumn("Order ID", width="large"),
            "restaurant_name": st.column_config.TextColumn("Restaurant Name", width="large"),
            "order_date": st.column_config.DatetimeColumn("Order Date", width="medium"),
            "order_value": st.column_config.NumberColumn("Order Value", format="%.2f", width="medium"),
            "discount_used": st.column_config.NumberColumn("Discount Used (1/0)", width="small"),
            "payment_method": st.column_config.TextColumn("Payment Method", width="medium"),
            "order_month_year": st.column_config.DatetimeColumn("Order Month/Year", format="YYYY-MM", width="medium")
        }
    )

    col3, col4 = st.columns([1, 1]) # Example of two equally sized columns for order data

    with col3:
        # Top 5 Restaurants with the maximum Order Value
        st.subheader("Top 5 Restaurants by Total Order Value")
        st.markdown("**Key Insights:**\n*   Highlights the restaurants that generate the highest total order value.\n*   These are key performers, and understanding their strategies could provide insights for other businesses.")
        top5_order_value = df_orders.groupby('restaurant_name')['order_value'].sum().nlargest(5).reset_index()
        st.dataframe(
            top5_order_value,
            column_config={
                "restaurant_name": st.column_config.TextColumn(
                    "Restaurant Name",
                    width="large"
                ),
                "order_value": st.column_config.NumberColumn(
                    "Total Order Value",
                    format="%.2f",
                    width="small"
                )
            }
        )

        # Monthly sales and discount usage
        st.subheader("Monthly Sales and Discount Usage")
        st.markdown("**Key Insights:**\n*   Analyzes sales trends and discount usage on a monthly basis.\n*   Helps identify peak sales months and assess the impact of discounts on overall revenue.")
        monthly_sales_and_discounts = df_orders.groupby('order_month_year').agg(
            total_sales=('order_value', 'sum'),
            total_discounts_used=('discount_used', 'sum')
        ).sort_values(by='total_sales', ascending=False)
        st.dataframe(
            monthly_sales_and_discounts,
            column_config={
                "order_month_year": st.column_config.DatetimeColumn("Month/Year", format="YYYY-MM", width="medium"),
                "total_sales": st.column_config.NumberColumn("Total Sales", format="%.2f", width="medium"),
                "total_discounts_used": st.column_config.NumberColumn("Total Discounts Used", width="medium")
            }
        )

        # Filtered Monthly Sales and Discounts (April 2025 onwards)
        st.subheader("Monthly Sales and Discounts from April 2025")
        st.markdown("**Key Insights:**\n*   This table displays the monthly sales and discounts, filtered from April 2025 onwards.\n*   It helps in analyzing recent trends and seasonal patterns without older data points.")
        start_date_period = pd.Timestamp('2025-04-01') # Changed from pd.Period
        filtered_monthly_sales_and_discounts = monthly_sales_and_discounts[monthly_sales_and_discounts.index >= start_date_period].sort_index(ascending=True)
        st.dataframe(
            filtered_monthly_sales_and_discounts,
            column_config={
                "order_month_year": st.column_config.DatetimeColumn("Month/Year", format="YYYY-MM", width="medium"),
                "total_sales": st.column_config.NumberColumn("Total Sales", format="%.2f", width="medium"),
                "total_discounts_used": st.column_config.NumberColumn("Total Discounts Used", width="medium")
            }
        )

    with col4:
        # Plotting the overall monthly sales trend
        st.subheader("Overall Monthly Sales Trend Visualization")
        fig_overall_monthly_sales = plt.figure(figsize=(12, 6))
        sns.lineplot(x=monthly_sales_and_discounts.index.astype(str), y='total_sales', data=monthly_sales_and_discounts, marker='o')
        plt.title('Overall Monthly Sales Trend')
        plt.xlabel('Month and Year')
        plt.ylabel('Total Sales')
        plt.xticks(rotation=45, ha='right')
        plt.grid(True, linestyle='--', alpha=0.7)
        plt.tight_layout()
        st.pyplot(fig_overall_monthly_sales)
        plt.close(fig_overall_monthly_sales)

        # Plotting filtered monthly sales over the last 6 months
        st.subheader("Monthly Sales Trend from April 2025 Visualization")
        fig_filtered_monthly_sales = plt.figure(figsize=(12, 6))
        sns.lineplot(x=filtered_monthly_sales_and_discounts.index.astype(str), y='total_sales', data=filtered_monthly_sales_and_discounts, marker='o')
        plt.title('Monthly Sales Trend (April 2025 onwards)')
        plt.xlabel('Month and Year')
        plt.ylabel('Total Sales')
        plt.xticks(rotation=45, ha='right')
        plt.grid(True, linestyle='--', alpha=0.7)
        plt.tight_layout()
        st.pyplot(fig_filtered_monthly_sales)
        plt.close(fig_filtered_monthly_sales)


        # Total orders over the last 6 months
        st.subheader("Total Orders Over the Last 6 Months")
        st.markdown("**Key Insights:**\n*   Provides a view of the total number of orders placed over the most recent six months.\n*   Useful for observing short-term trends and seasonality in customer activity.")
        latest_month = df_orders['order_month_year'].max()
        start_month = latest_month - pd.DateOffset(months=5) # Use DateOffset for datetime objects
        df_last_6_months = df_orders[df_orders['order_month_year'] >= start_month]
        orders_last_6_months = df_last_6_months.groupby('order_month_year').size().reset_index(name='total_orders')
        st.dataframe(
            orders_last_6_months,
            column_config={
                "order_month_year": st.column_config.DatetimeColumn("Month/Year", format="YYYY-MM", width="medium"),
                "total_orders": st.column_config.NumberColumn("Total Orders", width="medium")
            }
        )

        # Plotting total orders over the last 6 months
        st.subheader("Total Orders Over the Last 6 Months Visualization")
        fig_6_months_orders = plt.figure(figsize=(10, 6))
        sns.barplot(x='order_month_year', y='total_orders', data=orders_last_6_months, palette='viridis', hue='order_month_year', legend=False)
        plt.title('Total Orders Over the Last 6 Months')
        plt.xlabel('Month and Year')
        plt.ylabel('Total Orders')
        plt.xticks(rotation=45, ha='right')
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.tight_layout()
        st.pyplot(fig_6_months_orders)
        plt.close(fig_6_months_orders)

        # Average order value over the last 3 months
        st.subheader("Average Order Value Over the Last 3 Months")
        latest_month = df_orders['order_month_year'].max()
        # Corrected fix: Use DateOffset for datetime objects
        start_month_3_months = latest_month - pd.DateOffset(months=2)
        df_last_3_months = df_orders[df_orders['order_month_year'] >= start_month_3_months]
        average_order_value_last_3_months = df_last_3_months.groupby('order_month_year')['order_value'].mean().round(2).reset_index(name='average_order_value')
        st.dataframe(
            average_order_value_last_3_months,
            column_config={
                "order_month_year": st.column_config.DatetimeColumn("Month/Year", format="YYYY-MM", width="medium"),
                "average_order_value": st.column_config.NumberColumn("Average Order Value", format="%.2f", width="medium")
            }
        )

        # Plotting average order value over the last 3 months
        st.subheader("Average Order Value Over the Last 3 Months Visualization")
        fig_3_months_avg_value = plt.figure(figsize=(8, 5))
        sns.barplot(x='order_month_year', y='average_order_value', data=average_order_value_last_3_months, palette='magma', hue='order_month_year', legend=False)
        plt.title('Average Order Value Over the Last 3 Months')
        plt.xlabel('Month and Year')
        plt.ylabel('Average Order Value')
        plt.xticks(rotation=45, ha='right')
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.tight_layout()
        st.pyplot(fig_3_months_avg_value)
        plt.close(fig_3_months_avg_value)

# Close the database connection
conn.close()
```
